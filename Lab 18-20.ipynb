```python
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats

age = [23, 23, 27, 27, 39, 41, 47, 49, 50,
       52, 53, 54, 56, 57, 58, 60, 61, 62]

fat = [9.5, 26.5, 7.8, 17.8, 31.4, 25.9, 27.4, 27.2, 31.2,
       34.6, 42.5, 28.8, 33.4, 30.2, 34.1, 32.9, 41.2, 35.7]

df = pd.DataFrame({
    "Age": age,
    "%Fat": fat
})

print("Data:")
print(df)

print("\nMean:")
print(df.mean())

print("\nMedian:")
print(df.median())

print("\nStandard Deviation:")
print(df.std())

plt.figure(figsize=(8, 5))
plt.boxplot([df["Age"], df["%Fat"]], tick_labels=["Age", "%Fat"])
plt.title("Boxplots of Age and %Fat")
plt.ylabel("Values")
plt.show()

plt.figure(figsize=(8, 5))
plt.scatter(df["Age"], df["%Fat"])
plt.xlabel("Age")
plt.ylabel("%Fat")
plt.title("Scatter Plot of Age vs %Fat")
plt.show()

plt.figure(figsize=(6, 5))
stats.probplot(df["Age"], dist="norm", plot=plt)
plt.title("Q-Q Plot for Age")
plt.show()

plt.figure(figsize=(6, 5))
stats.probplot(df["%Fat"], dist="norm", plot=plt)
plt.title("Q-Q Plot for %Fat")
plt.show()
```

```python
import pandas as pd

data = {
    "Date": ["2026-01-01", "2026-01-02", "2026-01-03", "2026-01-04", "2026-01-05",
             "2026-01-06", "2026-01-07", "2026-01-08", "2026-01-09", "2026-01-10"],
    "Product": ["Laptop", "Phone", "Laptop", "Tablet", "Phone",
                "Headphones", "Tablet", "Laptop", "Phone", "Headphones"],
    "Quantity Sold": [5, 10, 3, 7, 8, 15, 5, 4, 12, 10],
    "Unit Price": [50000, 20000, 50000, 30000, 20000,
                   3000, 30000, 50000, 20000, 3000]
}

sales = pd.DataFrame(data)

sales.to_csv("sales_data.csv", index=False)

df = pd.read_csv("sales_data.csv")

print("Sales Data:")
print(df)

df["Total Sales"] = df["Quantity Sold"] * df["Unit Price"]

print("\nData with Total Sales:")
print(df)

product_sales = df.groupby("Product")["Total Sales"].sum()

product_profit = product_sales * 0.20

print("\nTotal Sales for Each Product:")
print(product_sales)

print("\nProfit for Each Product:")
print(product_profit)

overall_sales = df["Total Sales"].sum()
overall_profit = overall_sales * 0.20

print("\nOverall Sales:", overall_sales)
print("Overall Profit:", overall_profit)

print("\nTop 5 Most Profitable Products:")
print(product_profit.sort_values(ascending=False).head(5))
```

```python
import pandas as pd

data = {
    "Customer ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Age": [25, 32, 28, 45, 38, 22, 50, 35, 29, 41],
    "Gender": ["Male", "Female", "Male", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],
    "Total Spending": [65000, 25000, 12000, 80000, 35000,
                       15000, 90000, 45000, 18000, 55000]
}

customers = pd.DataFrame(data)

customers.to_csv("customer_data.csv", index=False)

df = pd.read_csv("customer_data.csv")

print("Customer Data:")
print(df)

def segment(spending):
    if spending >= 50000:
        return "High Spenders"
    elif spending >= 20000:
        return "Medium Spenders"
    else:
        return "Low Spenders"

df["Segment"] = df["Total Spending"].apply(segment)

print("\nCustomer Segmentation:")
print(df)

average_spending = df.groupby("Segment")["Total Spending"].mean()

print("\nAverage Spending in Each Segment:")
print(average_spending)

customer_count = df["Segment"].value_counts()

print("\nNumber of Customers in Each Segment:")
print(customer_count)
```
