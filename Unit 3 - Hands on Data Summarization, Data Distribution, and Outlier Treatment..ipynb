```python
import pandas as pd

data = {
    "Emp_ID":[101,102,103,104,105],
    "Name":["Arun","Bala","Chitra","Divya","Ezhil"],
    "Age":[25,30,28,35,29],
    "Salary":[30000,45000,40000,50000,55000],
    "Department":["HR","IT","Finance","IT","HR"]
}

df = pd.DataFrame(data)

print("Number of Rows:",df.shape[0])
print("Number of Columns:",df.shape[1])

print("\nData Types")
print(df.dtypes)

print("\nMissing Values")
print(df.isnull().sum())

print("\nStatistical Summary")
print(df.describe(include='all'))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create Student Grade Dataset
data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J"],
    "Grade": ["A", "B", "A", "C", "B", "A", "D", "C", "B", "A"]
}

df = pd.DataFrame(data)

# Calculate Frequency Distribution
frequency = df["Grade"].value_counts().sort_index()

print("Frequency Distribution of Grades")
print(frequency)

# Plot Bar Chart
plt.figure(figsize=(6,4))
plt.bar(frequency.index, frequency.values)

plt.title("Frequency Distribution of Students' Grades")
plt.xlabel("Grades")
plt.ylabel("Frequency")

plt.show()
```

```python
import pandas as pd

marks=[35,40,45,50,55,60,65,70,75,95]

df=pd.DataFrame({"Marks":marks})

Q1=df["Marks"].quantile(0.25)
Q3=df["Marks"].quantile(0.75)
IQR=Q3-Q1

lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

outliers=df[(df["Marks"]<lower)|(df["Marks"]>upper)]

print(outliers)
```

```python
import pandas as pd

sales=[100,120,130,140,150,160,170,180,1000]

df=pd.DataFrame({"Sales":sales})

Q1=df["Sales"].quantile(0.25)
Q3=df["Sales"].quantile(0.75)
IQR=Q3-Q1

lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

clean=df[(df["Sales"]>=lower)&(df["Sales"]<=upper)]

print(clean)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data=[10,12,13,14,15,16,18,20,100]

df=pd.DataFrame({"Value":data})

Q1=df["Value"].quantile(0.25)
Q3=df["Value"].quantile(0.75)
IQR=Q3-Q1

lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

clean=df[(df["Value"]>=lower)&(df["Value"]<=upper)]

plt.hist(df["Value"])
plt.title("Histogram Before Removing Outliers")
plt.show()

plt.boxplot(df["Value"])
plt.title("Box Plot Before Removing Outliers")
plt.show()

plt.hist(clean["Value"])
plt.title("Histogram After Removing Outliers")
plt.show()

plt.boxplot(clean["Value"])
plt.title("Box Plot After Removing Outliers")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data={
    "Age":[22,25,28,30,35,40,100],
    "Salary":[25000,30000,35000,40000,45000,50000,200000]
}

df=pd.DataFrame(data)

print(df.info())
print(df.isnull().sum())
print(df.describe())

df.hist(figsize=(6,4))
plt.show()

df.boxplot()
plt.show()

Q1=df.quantile(0.25)
Q3=df.quantile(0.75)
IQR=Q3-Q1

lower=Q1-1.5*IQR
upper=Q3+1.5*IQR

clean=df[~((df<lower)|(df>upper)).any(axis=1)]

print(clean)

clean.to_csv("cleaned_dataset.csv",index=False)

print("Dataset saved successfully.")
```
