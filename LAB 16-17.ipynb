```python
from collections import Counter
import string

reviews = [
    "This product is excellent",
    "Excellent quality and excellent service",
    "Good product and good quality",
    "Very good product"
]

text = " ".join(reviews).lower()

text = text.translate(str.maketrans('', '', string.punctuation))

words = text.split()

frequency = Counter(words)

print("Word Frequency Distribution")

for word, count in frequency.items():
    print(word, ":", count)
```

```python
import pandas as pd

data = {
    "feedback":[
        "The product is excellent",
        "Excellent service and good quality",
        "Good product and fast delivery",
        "Very good product"
    ]
}

df = pd.DataFrame(data)

df.to_csv("data.csv",index=False)

print("CSV file created successfully.")
```

```python
import pandas as pd
import string
from collections import Counter
import matplotlib.pyplot as plt

df = pd.read_csv("data.csv")

text = " ".join(df["feedback"]).lower()

text = text.translate(str.maketrans('', '', string.punctuation))

stop_words = {
    "the","and","is","a","an","of","to","in","on","for","with"
}

words = [word for word in text.split() if word not in stop_words]

frequency = Counter(words)

n = int(input("Enter Top N Words: "))

top_words = frequency.most_common(n)

print("\nTop", n, "Words")

for word, count in top_words:
    print(word, ":", count)

labels = [word for word, count in top_words]
values = [count for word, count in top_words]

plt.bar(labels, values)
plt.title("Top Frequent Words")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.show()
```
